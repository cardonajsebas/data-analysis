# 02. Build SQLite Database

This notebook is an optional step if the user want to avoid using sqlite in CLI to create world_indicators.db with the main schema and tables that will be used in the project

```python
import sqlite3
from pathlib import Path
```

```python
db_path = Path("world_indicators.db")
schema_path = Path("sql/schema.sql")

with sqlite3.connect(db_path) as conn:
    with open(schema_path, "r", encoding="utf-8") as f:
        sql = f.read()
    conn.executescript(sql)
```

```python
# Sanity check:
cur = conn.execute("SELECT name FROM sqlite_master WHERE type='table'")
print(cur.fetchall())
```
